## Graphique E en fonction de h

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.tan(x)*np.cosh(x)

def fp_exact(x):
    return (1/np.cos(x)**2)*np.cosh(x) + np.tan(x)*np.sinh(x)

N=[1000,2000,3000,5000,10000,200000]

E1=[]
E2=[]
E4=[]
h_liste=[]
eps=1e-2

for n in N:
    x,h=np.linspace(-np.pi/2+eps,np.pi/2-eps,n,retstep=True)
    y=np.tan(x)*np.cosh(x)
    h_liste.append(h)
    
    Eh=[]
    Eh2=[]
    Eh4=[]
    #Exact
    E=fp_exact(x)
    
    #Ordre h
    for i in range(0,len(x)-1): #on itère sur chaque point
        y_p=(y[i+1]-y[i])/(h) #calcul de la pente/dérivé
        Eh.append(abs(E[i]-y_p)/E[i]) #On ajoute la valeur dans une liste vide
    E1.append(np.max(Eh))
    
    #Ordre h2             
    for j in range(1,len(x)-1): #on itère sur chaque point
        y_p=(y[j+1]-y[j-1])/(2*(h)) #calcul de la pente/dérivé
        Eh2.append(abs(E[j]-y_p)/E[j]) #On ajoute la valeur dans une liste vide  
    E2.append(np.max(Eh2))
    
    #Ordre h4
    for k in range(2,len(x)-2): #on itère sur chaque point
        y_p=(y[k-2]-8*y[k-1]+8*y[k+1]-y[k+2])/(12*(h)) #calcul de la pente/dérivé
        Eh4.append(abs(E[k]-y_p)/E[k]) #On ajoute la valeur dans une liste vide  
    E4.append(np.max(Eh4))     

p1=np.polyfit(np.log(h_liste),np.log(E1),1)
p2=np.polyfit(np.log(h_liste),np.log(E2),1)
p4=np.polyfit(np.log(h_liste),np.log(E4),1)

print('pente ordre 1 =',p1[0])
print('pente ordre 2 =',p2[0])
print('pente ordre 4 =',p4[0])

print(len(E1))

plt.loglog(h_liste,E1, label="Erreur de la dérivation avec erreur en h",marker='.')
plt.loglog(h_liste,E2,label="Erreur de la dérivation avec erreur en h²",marker='.')
plt.loglog(h_liste,E4,label="Erreur de la dérivation avec erreur en h⁴",marker='.')
plt.title('Erreur en fonction du pas')
plt.ylabel("Erreur (max|f'num - f'exact|)")
plt.xlabel("Pas h")
plt.legend()
plt.savefig('Graph 1.2.png')
plt.show()
```

## Intégration par trapèze

### Code

```python
import numpy as np

def Int_trapèze(f,a,b,N):
    x,h=np.linspace(a,b,N,retstep=True)
    y=f(x)
    return np.sum(h/2*(y[:-1]+y[1:])) #Index [:-1] veut dire qu'on commence de [0] et on s'arrête au dernier point exclu et Index [1:] veut dire qu'on commence de [1] jusqu'au dernier point
        
```

## Intégration adaptative par Simpson

### Code

```python
import numpy as np
def Int_Simpson_adap(f,a,b,err):
    def Int_Simpson(f,a,b): #Sur 1 seul intervalle (a=xi-h, b=xi+h, xi est le milieu, h est le pas et f est la fonction à intégrer)
        milieu=(a+b)/2 #calcul de xi
        h=np.abs((b-a)/2)
        return (h/3*(F(a)+4*F(milieu)+F(b)))
            
    def F(x): #Compteur d'itération
        compteur["n_eval"] += 1
        return f(x)
        
    global compteur #On crée une variable globale qui servira de compteur 
    compteur={"n_eval": 0}
    
    total=Int_Simpson(f,a,b)
    stack=[(a,b,err,total)]

    rep=0
    Int=0
    while stack: #La boucle s'arrête quand stack devient vide
        a,b,err_loc,tot_loc=stack.pop()

        milieu=(a+b)/2
        gauche=Int_Simpson(f,a,milieu)
        droite=Int_Simpson(f,milieu,b)

        if abs(gauche + droite - tot_loc) <= 15*err_loc:
            Int += droite+gauche+(droite+gauche-tot_loc)/15 

        else:
            stack.append((milieu,b,err_loc/2,droite))
            stack.append((a,milieu,err_loc/2,gauche))
    return Int, compteur["n_eval"]
```

### Vérification

```python
def f_test(x):
    return x
    
I,n=Int_Simpson_adap(f_test,0,1,1e-9)

print("I =", I)
print("erreur =", abs(I - 0.5))
print("n_eval =", n)
```

## Runge-Kutta

```python
import numpy as np
import matplotlib.pyplot as plt

def R_Kutta4(f,t,y,dt,params): #le tau des notes de cours sera dt dans le code et f est la fonction dérivée qu'on connaît
    c1=dt*f(y,t,params)
    c2=dt*f(y+c1/2,t+dt/2,params)
    c3=dt*f(y+c2/2,t+dt/2,params)
    c4=dt*f(y+c3,t+dt,params)
    return y+1/6*(c1+2*c2+2*c3+c4) #Nous donne y_(i+1)

def Res_RK(f,y0,t0,t1,dt,params):
    n=int(np.ceil(abs(t1-t0)/dt))
    t=t0+dt*np.arange(n+1) #On crée un array de t0 à t1 par pas de n*dt
    y=np.zeros((n+1,len(y0))) #Dans le cas où y0 aurait plus que 1 dimension
    y[0]=y0
    
    for i in range(n):
        y[i+1]=R_Kutta4(f,t[i],y[i],dt,params)
    
    return y,t
          
```

```python
def fonction(Y,t,params):
    x, y, z = Y
    a = params["a"]
    return np.array([y + z, -x + a*y, x**2 - z], dtype=float)

def pente_log_distance(t, d, tmin=20, tmax=60):
    mask = (t >= tmin) & (t <= tmax) & np.isfinite(d) & (d > 0)
    tt = t[mask]
    yy = np.log(d[mask])
    if len(tt) < 10:
        return np.nan
    m, b = np.polyfit(tt, yy, 1) #On trouve le pente de la distance des 2 points, ce qui nous permet de savoir si le systèmes est chaotique
    return m

#paramètres initiaux
eps=1e-6

y0_1=np.array([0.1,0,0])
y0_2=np.array([0.1+eps,0,0])

t0,t1=0,100
dt=0.01

a_vals=np.linspace(-1,2,100)
mesure=[]

for a in a_vals:
    params={"a": float(a)}
    
    Y1,t_vec = Res_RK(fonction,y0_1,t0,t1,dt,params)
    Y2,_ = Res_RK(fonction,y0_2,t0,t1,dt,params)

    d = np.linalg.norm(Y2 - Y1, axis=1) #La distance entre 2 points est la norme du vecteur de séparation
    m = pente_log_distance(t_vec, d, tmin=30, tmax=80)  
    mesure.append(m)

mesure=np.array(mesure)

plt.plot(a_vals, mesure, marker=".", color='k')
plt.axhline(0)
plt.xlabel("a")
plt.ylabel("pente de log(distance)")
plt.title("Indicateur de chaos (pente de log(distance) en fonction de a)")
plt.grid()
plt.ylim(-0.5,0.3)
plt.xlim(-1,0.75)
plt.savefig('3.1.png')
plt.show()    


    
```

```python
import numpy as np

def sécante(f,x0,x1): #Trouver une racine par la méthode des sécantes
    tolérance=1e-12 #Plus haut que ça et les réponses sont fausses
    f0=f(x0)
    f1=f(x1)
    
    while True:
        denom=f1-f0
        if denom==0: #Puisqu'on divise par f1-f0, nous créons une pertubation si on se retrouve à diviser par 0
            x1+=1e-12
            f1=f(x1)
           
            denom=f1-f0
        x2=x1-f1*(x1-x0)/denom
        f2=f(x2)
        
        if abs(x2-x1)<tolérance or abs(f2) < tolérance:
            return x2
        x0,f0=x1,f1
        x1,f1=x2,f2
```
